## Analisis de datos


```python
path_file = os.path.join(DATA_DIR, MEASURE_DIR, files[0])
df = pd.read_csv(path_file, delimiter="\t", header=1)
df["Magnitude (dB)"].max()
df
```

## Graficar resultados preliminares


```python
#Grafico rapido del dataframe, frecuencia vs amplitud
x_ticks = [20, 100, 250,500,1000,2000,5000,10000,20000]
x_label = ['20', '100', '250', '500', '1000', '2000', '5000', '10000', '20000']
y_ticks = np.arange(-12, 13, 3)
plt.figure(figsize=(10, 6))
plt.plot(df['Frequency (Hz)'], df['Magnitude (dB)'])
plt.title('Respuesta en frecuencia del microfono Behringher ECM8000', fontsize=16)     
plt.xscale('log')
plt.xlim(20, 20000)
plt.ylim(-12, 12)
plt.yticks(y_ticks)
plt.xticks(x_ticks, x_label)
plt.xlabel('Frecuencia (Hz)', fontsize=12, fontname='Arial')
plt.ylabel('Nivel (dB)', fontsize=12, fontname='Arial') 
plt.grid(which='both', linestyle='--', linewidth=0.5)       
plt.savefig(os.path.join(output_path, "Behringher_0.png"), dpi=300)
plt.show()                        
```
